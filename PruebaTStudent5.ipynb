```python
import numpy as np
from scipy import stats
```

```python
# 1. Datos del ejercicio
pesos = [495, 498, 502, 490, 497, 499, 501, 493, 496, 498, 492, 495, 500, 494, 496]
media_referencia = 500
```

```python
# 2. Realizar la prueba t de una muestra
# ttest_1samp compara la muestra contra el valor esperado
t_stat, p_valor = stats.ttest_1samp(pesos, media_referencia)
```

```python
# 3. Cálculos descriptivos para el reporte
media_muestra = np.mean(pesos)
desviacion_std = np.std(pesos, ddof=1)
```

```python
# 4. Mostrar resultados
print("--- Resultados de la Prueba T de Una Muestra ---")
print(f"Media de la muestra: {media_muestra:.2f} g")
print(f"Desviación estándar: {desviacion_std:.2f} g")
print(f"Estadístico t: {t_stat:.4f}")
print(f"P-valor: {p_valor:.6f}")
```

```python
# 5. Interpretación
alpha = 0.05
if p_valor < alpha:
    print("\nConclusión: El p-valor es menor que 0.05.")
    print("SÍ hay evidencia suficiente para decir que el peso es DIFERENTE a 500g.")
    print("El inspector tiene razón: los sacos no están pesando lo que afirma la fábrica.")
else:
    print("\nConclusión: El p-valor es mayor que 0.05.")
    print("NO hay evidencia suficiente para rechazar la afirmación de la fábrica.")
```
