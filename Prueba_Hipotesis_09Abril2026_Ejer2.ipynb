```python
import numpy as np
from scipy import stats
```

EJERCICIO 2
Un investigador educativo quiere saber si existe diferencia en las calificaciones finales entre estudiantes que estudian con música y estudiantes que estudian en silencio.

Selecciona dos grupos completamente distintos y registra sus calificaciones:

Grupo Música: [65, 70, 68, 72, 66, 69, 71, 67, 70, 68]
Grupo Silencio: [85, 88, 90, 87, 92, 86, 89, 91, 88, 90]

Instrucciones:
Define la hipótesis nula y la hipótesis alternativa
Carga ambos grupos en arreglos de NumPy separados.
Aplica la prueba correcta de SciPy. Usa equal_var=False (prueba de Welch).
Usa α=0.01 en lugar del 0.05 habitual.
Extrae t y p-value, toma una decisión y redacta la conclusión. Recuerda que aquí los grupos son distintos (no son las mismas personas medidas dos veces).


```python
# Carga de datos en arreglos de NumPy
musica = np.array([65, 70, 68, 72, 66, 69, 71, 67, 70, 68])
silencio = np.array([85, 88, 90, 87, 92, 86, 89, 91, 88, 90])
```

```python
# Configuración del nivel de significancia
alpha = 0.01
```

```python
# Aplicar la prueba t de 2 muestras independientes (Welch)
t_stat, p_value = stats.ttest_ind(musica, silencio, equal_var=False)
```

```python
print(f"Estadístico t: {t_stat:.4f}")
print(f"p-value: {p_value:.4e}") # Usamos notación científica por ser muy pequeño
```

```python
# Toma de decisión
if p_value < alpha:
    print(f"Decisión: Rechazamos la hipótesis nula (H0) porque {p_value:.4e} < {alpha}")
else:
    print(f"Decisión: No hay evidencia suficiente para rechazar H0.")
```
