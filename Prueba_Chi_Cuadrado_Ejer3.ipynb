Ejercicio 3:
Un nutriólogo quiere saber si el tipo de dieta que sigue una persona está relacionado con su nivel de colesterol. Evalúa a 180 pacientes y registra:

                         Colesterol Alto    Colesterol Normal
Dieta vegana    10    50
Dieta mixta    30    40
Dieta carnívora    45    5
¿Existe relación entre el tipo de dieta y el nivel de colesterol? Usa
𝛼 = 0.05

```python
import numpy as np
from scipy import stats
```

```python
#Tabla de frecuencias observadas
tabla = np.array([
    [10, 50],
    [30, 40],
    [45, 5]
])
```

```python
#Realiza la prueba de chi cuadrado de independencia stats.chi2_contingency devuelve el esradistico chi-cuadrado, el p-value, los grados de libertad y las frecuencias esperadas
```

```python
chi2, p_value, g1, esperados = stats.chi2_contingency(tabla, correction=False)
```

```python
#Imprimir resultados
print(f"Estadístico chi-cuadrado= {chi2:.4f}")
print(f"P-value= {p_value:.4f}")
print(f"Grados de libertad = {g1}")
print("Frecuencias esperadas:")
print(esperados)

alpha = 0.05

if p_value < alpha:
    print("Hay evidencia suficiente para determinar relación entre el tipo de dieta y el nivel de colesterol.")
else:
    print("No hay evidencia suficiente para determinar relación entre el tipo de dieta y el nivel de colesterol.")
```
