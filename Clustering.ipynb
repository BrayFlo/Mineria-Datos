```python
import numpy as np #Biblioteca para calculos numericos
import matplotlib.pyplot as plt #Para graficar
from sklearn.cluster import KMeans # Implementacion de K'Means en sklearn
```

```python
#Definir los puntos de datos (nuestro conjunto de datos)
X = np.array([[1,2], [2, 3], [3, 3], [6, 8], [7, 9], [8, 8]])
```

```python
#Crear el modelo K-Means con K=2 (2 clusters)
Kmeans = KMeans(n_clusters=2, random_state=42) #random_state se utiliza para resultados reproducibles

#Ajustar el modelo a nuestros datos
Kmeans.fit(X)
```

```python
#Obtener las coordenadas de los centroides
centroides = Kmeans.cluster_centers_

#Obtener las etiquetas asignadasa cada punto de datos
etiquetas = Kmeans.labels_

#Imprimir los resultados
print("Centroides:", centroides)
print("Etiquetas:", etiquetas)
```

```python
#Graficar los puntos y los centroides
#X[:,0] significa que "toma todas las filas (:) de la columna 0 (las coordenadas X)"
#X[:,1] significa que "toma todas las filas (:) de la columna 1 (las coordenadas Y)"
plt.scatter(X[:,0], X[:,1], label='Puntos')
plt.scatter(centroides[:, 0], centroides[:, 1], c='red', marker='x', s=200, label='Centroides')
plt.xlabel('Eje X')
plt.ylabel('Eje Y')
plt.title('Clustering con K-Means')
plt.legend()
plt.show()
```
