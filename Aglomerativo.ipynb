```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram
from sklearn.cluster import AgglomerativeClustering
```

```python
X = np.array([
    [1, 1], #A
    [2, 1], #B
    [5, 5], #C
    [6, 5], #D
    [10, 10], #E
    [11, 10] #F
])

#Etiquetas para identificar puntos en las graficas
nombres = ["A", "B", "C", "D", "E", "F"]
```

```python
#3 Visualizar los datos originales

#Antes de aplicar el algoritmo, observamos los puntos
plt.scatter(
    X[:, 0], #todas las filas de la columna x
    X[:, 1] #todas las filas de la columna y
    )

#Agregamos nombre a cada punto
for nombre, (x, y) in zip(nombres, X): #zip() combina dos listas en una sola
    plt.annotate(nombre, (x, y)) #annotate() agrega texto a un punto especifico de la tabla

plt.title("Datos originales")
plt.xlabel("Coordenada X")
plt.ylabel("Coordenada Y")
plt.grid(True)

plt.show()
```

```python
#4 Construir la jerarquia (linkage)

Z =linkage(
    X,
    method = 'ward',
    metric = 'euclidean'
)
```

```python
#5 Explicar matriz linkage

#La matriz z contiene:
#[cluster 1, cluster 2, distancia, cantidad_elementos]

print("Matriz linkage:\n")
print(Z)
#
```

```python
#6 Dendrograma

dendrogram(
    Z,

    #Reemplazar indices 0,1,2 por A,B,C
    labels = nombres,

)

plt.title("Dendrograma del clustering jerarquico")
plt.xlabel("Puntos")
plt.ylabel("Distancia")

plt.grid(True)

plt.show()
```

```python
#7  Crear el modelo aglomerativo
modelo = AgglomerativeClustering(
    n_clusters=3,
    linkage='ward'
)
```

```python
#8 Entrenar el modelo
etiquetas = modelo.fit_predict(X)
```

```python
#9 Graficar resultado final
plt.scatter(
    X[:, 0],
    X[:, 1],
    c=etiquetas
)

for nombre, (x, y) in zip(nombres, X):
    plt.annotate(nombre, (x, y))

plt.title("Resultado del Clustering Jerarquico Aglomerativo")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)

plt.show()
```
