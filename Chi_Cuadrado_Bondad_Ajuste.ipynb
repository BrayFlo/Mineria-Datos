```python
import numpy as np
from scipy import stats
```

```python
#1 Frecuencias observadas (encuesta nueva)
observadas = np.array([90, 70, 40])
```

```python
#2 Frecuencias historicas esperadas
p = np.array([0.50, 0.30, 0.20])
```

```python
#3 Total de personas encuestadas
n = observadas.sum()
```

```python
#4 Frecuencias esperadas
esperadas = n * p
```

```python
#5 Calculo del estadistico chi-cuadrado
res = stats.chisquare(f_obs= observadas, f_exp= esperadas)
```

```python
#6 Extraer resultados
chi2 = res.statistic
p_value = res.pvalue
gl = len(observadas) - 1
```

```python
#7 Mostrar resultados
print(f"Frecuencias observadas: {observadas}")
print(f"Frecuencias esperadas: {esperadas}")
print(f"Chi-cuadrado: {chi2:.4f}")
print(f"Grados de libertad: {gl}")
print(f"Valor p: {p_value:.4f}")
```

```python
#8 Interpretacion
alpha = 0.05
if p_value < alpha:
  print("Se rechaza la hipotesis nula")
else:
  print("No se rechaza la hipotesis nula")
```
