Ejercicio 1: Ejercicio 1:
En una encuesta realizada en dos ciudades (Ciudad A y Ciudad B), se preguntó a los habitantes cuál era su medio de transporte preferido: automóvil o transporte público.

Los resultados fueron:
               Automóvil    Transporte Público
Ciudad A    85    65
Ciudad B    50    100

¿Existe una relación significativa entre la ciudad de residencia y la preferencia de transporte? Usa:
𝛼 = 0.05

```python
import numpy as np
from scipy import stats
```

```python
#Tabla de frecuencias observadas
tabla = np.array([
    [85, 65],
    [50, 100]
])
```

```python
#Realiza la prueba de chi cuadrado de independencia stats.chi2_contingency devuelve el esradistico chi-cuadrado, el p-value, los grados de libertad y las frecuencias esperadas
```

```python
chi2, p_value, g1, esperados = stats.chi2_contingency(tabla, correction=False)
```

```python
#Imprimir resultados
print(f"Estadístico chi-cuadrado= {chi2:.4f}")
print(f"P-value= {p_value:.4f}")
print(f"Grados de libertad = {g1}")
print("Frecuencias esperadas:")
print(esperados)

alpha = 0.05

if p_value < alpha:
    print("Hay evidencia suficiente para determinar la preferencia de transporte entre las ciudades de residencia.")
else:
    print("No hay evidencia suficiente para determinar la preferencia de transporte entre las ciudades de residencia.")
```
