EJERCICIO 3
Una universidad registra históricamente que sus estudiantes se distribuyen por carrera de la siguiente manera:

Ingeniería: 40%
Administración: 35%
Psicología: 25%
Este año se inscribieron 400 estudiantes nuevos y se observó la siguiente distribución: [200, 120, 80] (Ingeniería, Administración, Psicología)

Instrucciones:
Define la hipótesis nula y la hipótesis alternativa
Carga los datos observados en un arreglo de NumPy.
Define las proporciones esperadas y calcula las frecuencias esperadas multiplicando por 𝑛
Aplica la prueba correcta de SciPy con α=0.05.
Calcula los grados de libertad
Redacta la conclusión. Recuerda que aquí tienes una sola lista de categorías, no una tabla de dos variables.

```python
import numpy as np
from scipy import stats
```

```python
# 1. Datos observados
observados = np.array([200, 120, 80])
n_total = 400
```

```python
# 2. Proporciones y frecuencias esperadas
proporciones_esp = np.array([0.40, 0.35, 0.25])
esperados = proporciones_esp * n_total  # [160, 140, 100]
```

```python
# 3. Aplicar la prueba Chi-cuadrado de Bondad de Ajuste
chi_stat, p_value = stats.chisquare(f_obs=observados, f_exp=esperados)
```

```python
# 4. Calcular Grados de Libertad (k - 1)
# Donde k es el número de categorías (3 carreras)
grados_libertad = len(observados) - 1
```

```python
print(f"Frecuencias esperadas: {esperados}")
print(f"Estadístico Chi-cuadrado: {chi_stat:.4f}")
print(f"Grados de libertad: {grados_libertad}")
print(f"p-value: {p_value:.4f}")
```

```python
# Toma de decisión
alpha = 0.05
if p_value < alpha:
    print("\nDecisión: Rechazamos la hipótesis nula (H0).")
else:
    print("\nDecisión: No hay evidencia para rechazar H0.")
```
