Ejercicio 2:
Un investigador de salud pública quiere saber si el nivel educativo influye en el consumo de tabaco. Encuesta a 200 personas y obtiene:
                       Fuma    No Fuma
Sin estudios    40    20
Bachillerato    35    45
Universidad    15    45
¿El nivel educativo está relacionado con el hábito de fumar? Usa
𝛼 = 0.05

```python
import numpy as np
from scipy import stats
```

```python
#Tabla de frecuencias observadas
tabla = np.array([
    [40, 40],
    [35, 45],
    [15, 45]
])
```

```python
#Realiza la prueba de chi cuadrado de independencia stats.chi2_contingency devuelve el esradistico chi-cuadrado, el p-value, los grados de libertad y las frecuencias esperadas
```

```python
chi2, p_value, g1, esperados = stats.chi2_contingency(tabla, correction=False)
```

```python
#Imprimir resultados
print(f"Estadístico chi-cuadrado= {chi2:.4f}")
print(f"P-value= {p_value:.4f}")
print(f"Grados de libertad = {g1}")
print("Frecuencias esperadas:")
print(esperados)

alpha = 0.05

if p_value < alpha:
    print("Hay evidencia suficiente para determinar que el nivel educativo está relacionado con el hábito de fumar.")
else:
    print("No hay evidencia suficiente para determinar que el nivel educativo está relacionado con el hábito de fumar.")
```
