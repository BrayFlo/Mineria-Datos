```python
import numpy as np
from scipy import stats
```

```python
# 1. Definir los datos (Listas originales)
antes = [15.2, 16.0, 14.8, 15.5, 17.1, 16.4, 15.9, 16.2, 15.0, 15.7]
despues = [15.4, 16.1, 14.9, 15.6, 17.0, 16.5, 16.0, 16.3, 15.2, 15.8]
```

```python
# 2. Realizar la prueba t para muestras dependientes (relacionadas)
t_stat, p_valor = stats.ttest_rel(despues, antes)
```

```python
# 3. Mostrar resultados
print(f"Estadístico t: {t_stat:.4f}")
print(f"P-valor: {p_valor:.4f}")
```

```python
# 4. Interpretación (Nivel de significancia alpha = 0.05)
alpha = 0.05
if p_valor < alpha:
    print("Resultado: Hay un crecimiento significativo (Rechazamos H0)")
else:
    print("Resultado: No hay evidencia de crecimiento significativo (No rechazamos H0)")
```
