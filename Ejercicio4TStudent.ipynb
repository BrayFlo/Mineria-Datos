```python
import numpy as np
from scipy import stats
```

```python
# 1. Generación de datos simulados
# Supongamos 20 jóvenes y 20 adultos mayores (Total 40)
np.random.seed(42) # Para que los resultados sean siempre los mismos
jovenes = np.random.normal(loc=250, scale=30, size=20)   # Media de 250ms
adultos_mayores = np.random.normal(loc=310, scale=35, size=20) # Media de 310ms
```

```python
# 2. Realizar la prueba t para muestras independientes
# equal_var=True asume que las varianzas de ambos grupos son similares
t_stat, p_valor = stats.ttest_ind(jovenes, adultos_mayores)
```

```python
# 3. Mostrar resultados
print("--- Resultados de la Prueba T de Muestras Independientes ---")
print(f"Media Jóvenes: {np.mean(jovenes):.2f} ms")
print(f"Media Adultos Mayores: {np.mean(adultos_mayores):.2f} ms")
print(f"Estadístico t: {t_stat:.4f}")
print(f"P-valor: {p_valor:.6f}")
```

```python
# 4. Interpretación
alpha = 0.05
if p_valor < alpha:
    print("\nConclusión: El p-valor es menor que 0.05.")
    print("SÍ hay una diferencia significativa en la velocidad de reacción.")
else:
    print("\nConclusión: El p-valor es mayor que 0.05.")
    print("NO hay evidencia suficiente para decir que las velocidades son diferentes.")
```
