```python
import numpy as np
from scipy import stats
```

```python
# Carga de datos
datos = np.array([498, 501, 499, 502, 500, 497, 503, 499, 501, 500])
valor_teorico = 500
alpha = 0.05
```

```python
# Aplicar la prueba t de 1 muestra
t_stat, p_value = stats.ttest_1samp(datos, valor_teorico)
```

```python
print(f"Estadístico t: {t_stat:.4f}")
print(f"p-value: {p_value:.4f}")
```

```python
# Toma de decisión
if p_value < alpha:
    print("Rechazamos la hipótesis nula (H0).")
else:
    print("No hay evidencia suficiente para rechazar la hipótesis nula (H0).")
```
